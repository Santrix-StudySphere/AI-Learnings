# Decision Tree Classifier And Random Forest
# Predict the income of a adult based on census data

-----------------------------------------------------------------------

```python
#Import liabraries
import pandas as pd
```

```python
#Read the dataset
data  = pd.read_csv('04+-+decisiontreeAdultIncome.csv')
```

```python
#Check for null values
data.isnull().sum(axis=0)
```

```python
#Check the datatypes  - Need to convert all the non-numeric rows to numeric

data.dtypes
```

```python
#Get to Dummy Variables

data_prep = pd.get_dummies(data, drop_first=True)
```

```python
#Create the X and Y variables
X = data_prep.iloc[:,:-1]
Y = data_prep.iloc[:,-1]
```

```python
# Train and test the dataset
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = \
    train_test_split(X,Y, test_size=0.3, random_state=1234, stratify=Y)
```

```python
# Decision Tree Classifier:

from sklearn.tree import DecisionTreeClassifier
dtc = DecisionTreeClassifier(random_state=1234)
dtc.fit(X_train,Y_train)

Y_predict1 = dtc.predict(X_test)
```

```python
#Evaluate the model of Decision Tree Classifier:
from sklearn.metrics import confusion_matrix
cm1 = confusion_matrix(Y_test,Y_predict1)
score1 = dtc.score(X_test,Y_test)
print(score1)
```

```python
# Random Forest Classifier:

from sklearn.ensemble import RandomForestClassifier
rfc = RandomForestClassifier(random_state=1234)
rfc.fit(X_train,Y_train)

Y_predict2 = rfc.predict(X_test)
```

```python
#Evaluate the model of Random Forest Classifier:

cm2 = confusion_matrix(Y_test,Y_predict2)
score2 = rfc.score(X_test,Y_test)
print(score2)
```
