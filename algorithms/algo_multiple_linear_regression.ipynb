```python
#Multiple Linear Regreassion
# One dependent and multiple independent variables

#import Pandas
import pandas as pd
```

```python
# Read the csv file 
dataset = pd.read_csv('02Students.csv')
#Create a data frame
df = dataset.copy()
#Divide the data into X and Y
X = df.iloc[:, :-1]
Y = df.iloc[:, -1]

# import train_test_split from sklearn.model_selection to split the dataset into rows as train and test data
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.3, random_state=1234)

# import linear regression model from sklearn.linear_model
from sklearn.linear_model import LinearRegression

#Create a regressor
std_reg = LinearRegression()

#Train the regressor model on the train data
std_reg.fit(x_train,y_train)

#Predict the y-value
y_predict = std_reg.predict(x_test)

#Find R-square value

mlr_score = std_reg.score(x_test,y_test)   #came approx 0.901

#Coefficents of the line:
mlr_coefficent = std_reg.coef_
mlr_intercept = std_reg.intercept_

#Therefore, the equation of the line becomes:
# y = 1.3152 + (4.6655)x1 + (5.0748)x2

# How much error our model has made
# Calculate the RMSE

#import mean_squared_error form sklearn.metrics
from sklearn.metrics import mean_squared_error
#import math
import math
mlr_rmse = math.sqrt(mean_squared_error(y_test, y_predict))   #comes approx 5.75536



```
