```python
# SIMPLE LINEAR REGRESSION USING PYTHON

#import Pandas
import pandas as pd
```

```python
# Read csv tto store students hours vs marks data in a dataset

dataset = pd.read_csv('01Students.csv')
df = dataset.copy()

# Split the data into X and Y:
X = df.iloc[:, :-1]
Y = df.iloc[:, -1]

# Split the dataset by rows into trainning and test data
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(X,Y,test_size=0.3,random_state=1234)

#Create Simple Linear Regression model
from sklearn.linear_model import LinearRegression

#Create regressor ->
#  In sklearn, a regressor is a model class (like LinearRegression, Ridge, etc.) that predicts continuous values,
#  as opposed to a classifier (which predicts categories).
std_reg = LinearRegression()

#Train or fit our trainning data 
std_reg.fit(x_train,y_train)

# Now predict the data of Y from the test data
y_predict = std_reg.predict(x_test)
```
