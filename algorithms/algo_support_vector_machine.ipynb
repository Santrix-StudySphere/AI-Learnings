```python
# Import liabraries
import pandas as pd
```

```python
#Read the data and create a copy
LoanData = pd.read_csv('01Exercise1.csv')

LoanPrep = LoanData.copy()
```

```python
#Find ot columns with missing values
#LoanPrep.isnull() -> All values in every row is replaced with 'True' for null value and 'False' for not-null value
#axis=0 means sume is carried out downwards accross each column.
#axis=1 means sum is carried out horizontal accross each roow
LoanPrep.isnull().sum(axis=0)
```

```python
#Replace the Missing values. Drop the rows
LoanPrep = LoanPrep.dropna()
```

```python
# Drop the irrelevent columns based on business sens
#Gender cannot be a predicting factors as Bank will not decide the approval on the basis of Gender. Hence this column can be dropped.

LoanPrep = LoanPrep.drop(['gender'],axis=1)
```

```python
#Create Dummy Variables
LoanPrep.dtypes   # checking the data types of the columns

LoanPrep = pd.get_dummies(LoanPrep, drop_first=True)     #Martital status and Approval status
```

```python
#Normalize the data for loanamt and income using StandardScaler
from sklearn.preprocessing import StandardScaler
scaler_ = StandardScaler()

LoanPrep['income'] = scaler_.fit_transform(LoanPrep[['income']])
LoanPrep['loanamt'] = scaler_.fit_transform(LoanPrep[['loanamt']])
```

```python
# Create the X (Independent) and Y (Dependent) dataframes
Y = LoanPrep[['status_Y']]
X = LoanPrep.drop(['status_Y'],axis=1)
```

```python
#8. Split the X and Y dataset into Train and Test data
from sklearn.model_selection import train_test_split

X_train,X_test,Y_train,Y_test = \
    train_test_split(X,Y,test_size=0.3, random_state=1234, stratify=Y)
```

```python
#Build the Support Vector Classifier

from sklearn.svm import SVC

svc = SVC()

svc.fit(X_train, Y_train)
```

```python
#Predict thr outcome using test data

Y_predict = svc.predict(X_test)
```

```python
#Build the confusion matrix and get the accuracy score
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(Y_test, Y_predict)

score = svc.score(X_test, Y_test)
print(score)
```
