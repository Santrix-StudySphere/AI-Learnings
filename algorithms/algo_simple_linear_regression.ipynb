```python
# SIMPLE LINEAR REGRESSION USING PYTHON

#import Pandas
import pandas as pd
```

```python
# Read csv tto store students hours vs marks data in a dataset

dataset = pd.read_csv('01Students.csv')
df = dataset.copy()

# Split the data into X and Y:
X = df.iloc[:, :-1]
Y = df.iloc[:, -1]

# Split the dataset by rows into trainning and test data
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(X,Y,test_size=0.3,random_state=1234)

#Create Simple Linear Regression model
from sklearn.linear_model import LinearRegression

#Create regressor ->
#  In sklearn, a regressor is a model class (like LinearRegression, Ridge, etc.) that predicts continuous values,
#  as opposed to a classifier (which predicts categories).
std_reg = LinearRegression()

#Train or fit our trainning data 
std_reg.fit(x_train,y_train)

# Now predict the data of Y from the test data
y_predict = std_reg.predict(x_test)
```

```python
#Calculate the R-square and the eqpuation of the line
#R-square(Coefficient of Determination): 
   #It is a statistical measure used in regression analysis to show how well the independent variable(s) explain the variability of the dependent variable.
   # It ranges from 0 to 1
   # 0 → The model explains none of the variability.
   # 1 → The model explains all the variability perfectly.
   #Formula:
            # R^2 = 1 - (SSR/SST)
            # SSR : Sum of squared residuals (errors between predicted and actual values).
            # SST : Total sum of squares (variation of actual values from their mean).

slr_score = std_reg.score(x_test,y_test)   # This provides the R^2 value; came approx 0.84452

# Coefficent of the line:
slr_coefficent = std_reg.coef_             #slope of the line
slr_intercept = std_reg.intercept_         # y-intercept

# Therefore, the equation of the line becomes:
# y = (5.02709483)x + 34.27245358755  or approx y = 5.03x + 34.3
```

```python
# Calculate how much error our model has made
# RMSE - Root Mean Squared Error
from sklearn.metrics import mean_squared_error
import math 

slr_rmse = math.sqrt(mean_squared_error(y_test, y_predict))  #comes approx 7.228
```

```python
# Plotting the reasults using matplotlib
import matplotlib.pyplot as plt

#Plot scatter plot for the test data
plt.scatter(x_test, y_test)

#Plot trendline of the prediction
plt.plot(x_test, y_predict)
plt.ylim(ymin=0)
plt.show()
```
