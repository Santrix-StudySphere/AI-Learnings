# Iris Plant Classification Using SVM
# Check the effect of different Kernel functions

```python
#Import data
from sklearn import datasets

iris = datasets.load_iris()
print(iris)
```

```python
#Extract the data as X and Y

X = iris.data
Y = iris.target
```

```python
#Split the X and Y dataset into train and test data

from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = \
        train_test_split(X, Y, test_size=0.3, random_state=1234, stratify=Y)
```

```python
# Implemet the SVC algorithm 
# Train the SVC

from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix

```

```python
# RBF Kernel with gamma as 1
svc = SVC(kernel='rbf', gamma=1.0)
svc.fit(X_train,Y_train)
Y_predict = svc.predict(X_test)
cm_rbf01 = confusion_matrix(Y_test,Y_predict)
```

```python
# RBF Kernel with gamma as 10
svc = SVC(kernel='rbf', gamma=10.0)
svc.fit(X_train,Y_train)
Y_predict = svc.predict(X_test)
cm_rbf10 = confusion_matrix(Y_test,Y_predict)

#Conclusion:
# gamma increased -> radius reduced hence some of the points went out of the radidus leading to decrese in the accurace.
```

```python
#Linear
#Polynomial
#Sigmoid 

#Linear 

svc = SVC(kernel='linear')
svc.fit(X_train,Y_train)
Y_predict = svc.predict(X_test)
cm_linear = confusion_matrix(Y_test,Y_predict)

#Polynomial

svc = SVC(kernel='poly')
svc.fit(X_train,Y_train)
Y_predict = svc.predict(X_test)
cm_poly = confusion_matrix(Y_test,Y_predict)

#Sigmoid - > Not useful for this dataset (conclusion from results)

svc = SVC(kernel='sigmoid')
svc.fit(X_train,Y_train)
Y_predict = svc.predict(X_test)
cm_sig = confusion_matrix(Y_test,Y_predict)
```
