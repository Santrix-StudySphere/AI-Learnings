```python
#----------------------------------DATA PROCESSING --------------------------------------------------->
# Importing Pandas liabrary:

import pandas as pd
```

```python
# read the csv file using pandas
dataset = pd.read_csv('loan_small.csv')

# Access the dataframe data using iloc: Getting first 3 roows from 2nd and 3rd columns
subset1 = dataset.iloc[0:3,1:3]

# Access the dataframe using column names:
subset2 = dataset[['Gender','ApplicantIncome']][0:3]

# read the TSV file using Pandas
datasetT = pd.read_csv('loan_small_tsv.txt',sep='\t')

## Functions of pandas:
## head - get a quick view of the dataset (by default it fetches first 5 rows)
dataset.head()

## Get the shape of the data set 
dataset.shape

## Get the column names of the dataset
dataset.columns

## Find out columns with missing values
dataset.isnull().sum(axis=0)

## Data Pre Processing by replacing missing values
# As ML Algorithms do not understand the null or missing values we need to deal with them prior
# Deleting the records with null values - This can be done if null records are less

clean_data = dataset.dropna()

# Deleting record for a particular column null value
clean_data = dataset.dropna(subset=['Loan_Status'])

# copoy the dataset to do processing
dt = dataset.copy()
# Replacig Catagorical values with mode (most frequently occuring values)
cols1 = ['Gender','Area','Loan_Status']
dt[cols1] = dt[cols1].fillna(dt.mode().iloc[0])      #dt.mode() gives the most frequent value and iloc is user to get the first value in case there are more than 1 mode
dt.isnull().sum(axis=0)

# Replacing numerical values with mean
cols2 = ['ApplicantIncome','CoapplicantIncome','LoanAmount']
dt[cols2] = dt[cols2].fillna(dt[cols2].mean())
dt.isnull().sum(axis=0)

# Label Encoding using pandas
dt.dtypes     # to check the data types of the columns

# Change the data types of object type to 'category' (A Pandas datatype for catagorical data)
dt[cols1] = dt[cols1].astype('category')

# Replacing the categorical values with numerical values (Label encoding)
for column in cols1:
    dt[column] = dt[column].cat.codes   

# One Hot Encoding (OHE):
#One-Hot Encoding is a technique used in machine learning and statistics to convert categorical variables (labels that are not numeric) into a numerical format that algorithms can understand.
#Instead of assigning numbers directly (like Label Encoding), 
# One-Hot Encoding creates binary (0/1) vectors for each category.

# Removing the LoanId column to perform OHE on only required catagorical columns
df2 = dataset.drop(['Loan_ID'],axis=1)
# Implementing OHE
df2 = pd.get_dummies(df2)   # get_dummies(dataset) creates the dummy columns
```
